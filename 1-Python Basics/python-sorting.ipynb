```python
##bubble sort
nums=[5,4,3,2,1]


for i in range(len(nums)):
           for j in range(len(nums)-1-i):
               if nums[j]>nums[j+1]:
                  nums[j],nums[j+1]= nums[j+1], nums[j]
       
print(nums)
```

```python
#insertion sort

nums= [5,4,3,2,1,0]

i=0
res=[nums[0]]

for j in range(1, len(nums)):
    elem= nums[j]  
    for k in range(len(res)):
        if elem<=res[k]:
            res.insert(k,elem)
            break

print(res)




```
