map

```python
numbers=[1,2,3,4,5]

list(map(lambda x:x**2, numbers))
```

convert list of string to integer

```python
nums=["12", "1","3","4", "ioo"]

list(map(lambda x: int(x) if x.isdigit() else "NaN", nums))

```
