```python
arr = [1, 2, 2, 3, 4, 4, 5, 1]

unique=list(set(arr))
print(unique)

unique=[]

for i in arr:
    if i not in unique:
        unique.append(i)

print(unique)

seen= set()
for i in arr:
    if i not in seen:
        seen.add(i)

unique= list(seen)
print(unique)
```
