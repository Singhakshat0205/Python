## unpacking with *


```python

numbers=(1,2,3,4,5)
num=[1,2,3]
print(num)
num=tuple(num)
print(num)
num=list(num)
print(num)

first, *middle, last= numbers

print(first)
print(tuple(middle))
print(last)
```

Dictionaries

```python
dict={'k1':1, 'k2':[1,2,3]}


print(dict['k2'])
print(dict.get('k1'))

dict['k4']=5

print(dict)
del dict['k2']
print(dict)
```

Dictionary Methods


```python
keys= dict.keys()
print(keys)

values= dict.values()
print(values)

items= dict.items()

print(items)
```

Shallow Copy

```python
student= {'name': "a", "age":19}

student_copy= student

student_copy['age']=20

print(student)
print(student_copy)


# how to do deep copy

student_copy_deep=student.copy()

student_copy_deep['age']=30

print(student)
print(student_copy_deep)
```

Iterating over dictionaries


```python
for key,value in student.items():
      print(f"{key}:{value}")
```

Dictionary Comprehension


```python
squares= {x:x**2 for x in range(5)}
print(squares)
```

conditional dictionary comprehension


```python
evens= {x: x**2 for x in range(10) if x%2==0}
print(evens)
```

use a dictionary to count the frequency of elements in the list

```python
numbers=[1,1,2,3,4,5,5,6,6,6,6,3,4,6,6,0,9]

dict_freq= {x:numbers.count(x) for x in numbers}

print(dict_freq)
```

merge two dictionaries

```python
dict1={'a':1, 'b':2}
dict2={'b':4, 'c':3}

merged_dict={**dict1, **dict2}
print(merged_dict)
```
