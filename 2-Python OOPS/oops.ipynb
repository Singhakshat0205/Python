```python
## instance variable and methods
class Dog():
    ## constructor
    def __init__(self, name, age):
          self.name=name
          self.age= age
    
    def bark(self):

        print(f"{self.name} says woof")

Dog1= Dog("bhaubhau", 3)
print(Dog1.name)
print(Dog1.age)
Dog1.bark()
```

inheritence


```python
class Car():

    def __init__(self, windows, doors ,enginetype):
        self.windows= windows
        self.doors= doors
        self.enginetype = enginetype
    
    def Drive(self):
        print("Car is running")
    
class Tesla(Car):

    def __init__(self, windows,doors, enginetype, selfdrive):
        super().__init__(windows, doors, enginetype)
        self.selfdrive= selfdrive



Tesla1 = Tesla(4,4,"hybrid", True)
print(Tesla1.windows)
c1= Car(1,2,"Self")

c1.Drive()

```

Abstract Methods


```python
class Animal():


    def sound(self):
        pass


class Dog(Animal):
    def sound(self):
        print("Bark")

Dog1= Dog()

Dog1.sound()
```

Encapsulation


```python
##private variable


class Person:

    def __init__(self,name, age):
        self.__name= name ##private variables
        self.__age= age

    def getter(self):

        name= self.__name
        age= self.__age
        return {"Name":name, "Age":age}
    
person1= Person("Akshat", 24)
person1.getter()

    
```
