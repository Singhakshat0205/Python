Here we will learn how we can use DataSet and Dataloader classes to implement the Mini Batch GD in our NN model

Dataset class has the responsibility to load the dataset from the memory

Dataloader class will make batches from the loaded data

__init__()-> tells how data should be loaded
__len__()-> which returns the total number of batches
__getitem__(index)-> returns the data at the given index

```python
import torch
```

```python
from sklearn.datasets import make_classification
```

```python
## step 1-> create a synthetic classification dataset using sklearn
X, y= make_classification(

    n_samples = 10,
    n_features= 2,
    n_informative=2,
    n_redundant=0,
    n_classes= 2,
    random_state= 42
)
```

```python
X
```

```python
X.shape
```

```python
#convert numpy arrays to pytoch tensors

X= torch.tensor(X, dtype=torch.float32)
y= torch.tensor(y, dtype= torch.long)
```

```python
from torch.utils.data import Dataset, DataLoader

```

```python
class CustomDataset(Dataset):

    def __init__(self, features, labels):

        self.features= features
        self.labels= labels
    
    def __len__(self):
        return self.features.shape[0]

    
    def __getitem__(self, index):
        return self.features[index], self.labels[index]
```

```python
dataset= CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataset[0]
```

```python
dataloader= DataLoader(dataset, batch_size=2, shuffle= True, num_workers= 2)

### Very important concept "collate_fn"
'''
the collate_fn is a function that specifies how to combine a list of samples from a dataset into a single batch. by default DataLoader uses a single batch collation mechanism but collate_fn allows you to customize how the data should be processed and batched
'''


## when shuffle is True then we are using random sampler to sample the data
## but if shufffle is false then the sampling is going to be sequential
# we can also make our custom samplers for sampling the data
```
