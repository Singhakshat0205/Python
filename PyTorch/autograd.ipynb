```python
'''
what is autograd

is core component of pytorch that provides automatic differentiation for tensor operations

'''
```

```python
'''
backward() performs reverse-mode automatic differentiation through the computation graph and computes gradients of the output with respect to tensors that have requires_grad=True. The resulting gradients are accumulated in the tensors' .grad attributes.'''
```

```python

import torch

x= torch.tensor(3.0, requires_grad= True)


y= x**2
print(y)
a= torch.rand(3,4, requires_grad=True)
c= a**3
print(x)
print(a)
print(c)
y.backward()
x.grad


```

```python
x= torch.tensor(3.0, requires_grad= True)
y= x**2
z= torch.sin(y)
print(z)
print(y)
print(x)
```

```python
print(z)
print(y)
print(x.grad)
```

```python
x= torch.tensor(10.0, requires_grad= True)
y= x**2
z= y**2

z.backward()
print(x.grad)
```
