```python
##Mathematical operations in pytorch

a= torch.arange(10)
c= a+2
print(a)
print(c)

d= a//2
print(d)

e= a**2
print(e)


## using tensor to tensor operations

a= torch.empty(2,3)
a=a*100
b= torch.ones(2,3)
print(a+b)

d= a-b
print(d)

e= a*b

print(e)

f= a//b
print(f)
```

```python
##tensor of given size

a= torch.randint(size=(2,3),low=0, high=20)

print(a)
```

```python
##sum of whole tensor

s= torch.sum(a)
print(s)

## row wise sum
a= torch.rand(2,6)*100

print(a)
s= torch.sum(a,dim=0)
print(s)

##column wise sum

s= torch.sum(a, dim=1)
print(s)

##Product 
print(torch.prod(a))
```

```python
##Mean, median and mode
a = torch.tensor([[1.0, 2.5, 3.0], 
                  [4.2, 5.0, 6.1]], dtype=torch.float32)

a= torch.rand(2,3)
## column mean
print(torch.mean(a,dim=1))

##row mean
print(torch.mean(a, dim=0))
```

```python
maxi= torch.max(a)
print(maxi)
mini= torch.min(a)
print(mini)
##below tells index of max element
print(torch.argmax(a))

print(torch.argmin(a))
```

```python
##Matrix multiplication
a=torch.rand(3,4)
b=torch.rand(4,3)

c= torch.matmul(a,b)
```

```python
## transpose
a= torch.rand(3,4)

print(a)
b= torch.transpose(a,0,1)
print(a)
print(b)
```

```python
##Determinent
a= torch.rand(4,4)
c= torch.det(a)
print(c)
```

```python
##comparison operatins
a= torch.rand(2,3)
b=torch.rand(2,3)
print(a>b)
```

```python
## sqrt

v= torch.sqrt(a)
print(v)
```

```python
a= torch.ones(3,4)

print(torch.sigmoid(a))
print(torch.softmax(a, dim=1))

print(torch.relu(a))
```

```python
##in-place opertions
##changing original tensors

a= torch.rand(10,10)
b= torch.rand(10,10)

a.add_(b)

print(a)
```

```python
print(a)
a.relu_()
print(a)
```

```python
###copying a tensor

##shallow copy
a= torch.rand(2,3)
b=a

##Deep copy
c= a.clone()
c[0][0]=1
print(c)
print("original was below")
print(a)


```

```python
##Reshape
a= torch.rand(10,10)

print(a)
a.reshape(10,2,5)

##does not modifies the original tensor gives a new tensor
print(a)
```
