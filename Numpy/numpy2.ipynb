Numpy arrays vs Python lists

```python
import numpy as np
```

```python
#normal indexing 

a= np.arange(12).reshape(3,4)

print(a)

print(a[1::,1::2])
```

```python
##fancy indexing

a= np.arange(24).reshape(6,4)
print(a)
# for rows 
print(a[[0,2, 5,4]])
print(a[[0,2,4]])
#for columns 
print(a[:,[0,1,3]])
```

```python
#boolean indexing
##generates 24 random integers between 1 to 100
a= np.random.randint(1,100,24).reshape(6,4)
print(a)
```

```python
#print all the numbers greater than 50
print(a>50)
res_array_1d=a[a>0]
print(res_array_1d)
#numbers greater than 50 and also even
a[(a>50) & (a%2==0)]
#numbers not divisible by 7

print(a[~(a%7==0)])
```

Broadcasting in numpy


```python
a=np.arange(6).reshape(2,3)
b= np.arange(3)
print(a+b)






```

```python
a= np.arange(16).reshape(4,4)
b= np.arange(4).reshape(2,2)

print(a+b)
```

(4, 3)(3,)4, 3   3✅ YES3 matches 3. The empty space acts like a 1.(4, 3)(4, 1)4, 34, 1✅ YES4 matches 4. The 1 stretches to match 3.(4, 4)(2, 2)4, 42, 2❌ NO4 and 2 do not match, and neither is 1.

The Mental Shortcut: "The Rubber Band Rule"Think of the number 1 as a rubber band.If a dimension is 1, NumPy can stretch it to any number to match the other array.If a dimension is missing (empty space on the left), NumPy automatically treats it as a 1 and stretches it.If a dimension is 2, 3, 4, etc., it is frozen. It cannot be stretched to match a different frozen number.


Working with mathematics

```python
a= np.arange(10)
print(np.sum(a))
print(np.sin(a))
```

```python
##sigmoid

def sigmoid(array):
    return 1/(1+np.exp(array))


a=np.arange(10)
b=sigmoid(a)
print(a)
print(b)
```

```python
#mean squared error

actual= np.random.randint(1,50,25)
predicted= np.random.randint(1,50,25)

#mse
def MSE(actual, predicted):
  return np.mean( (actual-predicted)**2)


MSE(actual, predicted)
```

```python
#binary cross entropy

np.mean((actual-predicted)**2)
```

```python
##working with missing values

a=np.array([1,2,3,np.nan,5])
print(a)

#finding missing values
a[~np.isnan(a)]
```
