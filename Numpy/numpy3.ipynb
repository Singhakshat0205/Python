Numpy Tricks 

```python
import numpy as np

a= np.arange(10)
np.sort(a)
b= np.random.randint(1,100,15).reshape(5,3)

c=np.sort(b, axis=0)[::-1]
print(c)
```

```python
a=np.arange(100)
c=np.append(a,200)
print(a)
print(c)
```

```python
d= np.append(b,np.ones((b.shape[0],1)), axis=1)
print(b)
print(d)
```

```python
## concatenate

a= np.arange(12).reshape(3,4)
b= np.arange(12).reshape(3,4)

c= np.concatenate((a,b),axis=0)
print(c)
```

```python
a= np.arange(12).reshape(3,4)
b= np.arange(15).reshape(3,5)
## row-wise concatenation below
np.concatenate((a,b), axis=1)
## column wise concatenation 
a= np.arange(12).reshape(3,4)
b=np.arange(20).reshape(5,4)

c= np.concatenate((a,b),axis=0)
print(c)
```

```python
## unique function

a=np.array([1,2,3,4,4,54,34])
c= np.unique(a)
print(c)

##for 2-D arrays
b= np.arange(12).reshape(3,4)
c= np.unique(b)
print(c)

```

```python
##np.expand_dims
#1-d=>2-D
a=np.arange(12)


## in numpy 0=>rows and 1=>columns
c=np.expand_dims(a,axis=1)
print(c)
```

```python
## Where

c= np.where(a>10)
print(c)
d= np.where(a%2==0)
print(d)
```

```python

##  find all the indices with value greater than 50
a= np.arange(100).reshape(10,10)
c= np.where(a>50)
print(c)
print(np.shape(c))
```

```python
## replace all values >50 with 0

np.where(a>50,0,a)

## replace evens with 0

np.where(a%2==0, 0, a)
```

```python
## argmax

## return the indices of the max element of the array in a particular axis
a= np.arange(100)
a[99]=1
a[98]=99
print(np.argmax(a))
print(a[np.argmax(a)])


a= np.arange(50).reshape(5,10)
print(a)
np.argmax(a, axis=0)#column wise max
np.argmax(a, axis=1)#columns wise max
 

 ## we also have argmin

print(np.argmin(a))

```

```python
np.cumsum(a)
print(a)
np.cumsum(a, axis=0)
np.cumsum(a, axis=1)
```

```python
np.cumprod(a)
np.cumprod(a, axis=1)
```

```python
np.percentile(a, 50)
```

```python


np.histogram(a, bins=10)
```

```python

a= np.arange(10)
b= np.random.randint(1,20,10)
np.corrcoef(a,b)

#  a b
# a
# b
```

```python
## np.isin

items= [1,20,30, 40,50,60, 70,80,90]
np.isin(a,items)

a[np.isin(a,items)]
```

```python
np.flip(a)
# reverses the array preserving the shape of the array

b= np.arange(100).reshape(10,10)
np.flip(b, axis=0) ## column wise
np.flip(b)## row wise + column wise 
np.flip(b, axis=1) # row wise
```

```python

a= np.arange(10)
b= np.delete(a,[0,2])
print(b)
```

```python
### set functions

np.union1d(a,b)
np.intersect1d(a,b)
np.setdiff1d(a,b)
np.setxor1d(a,b)
np.in1d(a,b)
```
