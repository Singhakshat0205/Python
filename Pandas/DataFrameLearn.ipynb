```python
import pandas as pd
```

```python
students= [

    [10,20,30],
    [20,33,34],
    [78,55,85],
    [44,55,78]


]


df= pd.DataFrame(students,columns=['Maths', 'Phy', 'Chem'])
print(df)
print(df.sample())


df.loc[df['Chem']%2==0, 'Chem']=df['Chem']+10
print(df)
```

```python
##how to change the values of a dataframe
print(df)
df.loc[0,"Maths"]=23

## we can also apply some condition while changing the values
df.loc[df['Phy']>10, 'Phy']=19
print(df)

## below is one more variant of updating the dataframe columns 
df.loc[df['Chem']%2==0, 'Chem']=df['Chem']+10
print(df)

## we can update wither one column or we can pass a list of the columns  we want to update and update all those columns in one go as below:
df.loc[df['Chem']>40, ['Maths', 'Chem', 'Phy']]=5500
print(df)
##changing value of a whole column 

# df['Maths']+=10
print(df)




```

```python
## we can also rename columns in the data frame
df.rename(
    columns={'Maths':'Mathematics'},
    inplace=True
)


## if we do not use the inplace= true
df= df.rename(
    columns={'Maths':'Mathematics',
    'Chem':'Chemistry',
    'Phy':'Physics'}
)
print(df)
```

```python
## Now we will learn how we can add new rows in our data frame

new_row= pd.DataFrame({
    'Mathematics':[99],
    'Physics':[89],
    'Chemistry':[99]
})

new_row.shape

df= pd.concat([df,new_row], ignore_index=True)
print(df)



```

```python
student_dict = {
    'name':['nitish','ankit','rupesh','rishabh','amit','ankur', 'ankur'],
    'iq':[100,90,120,80,0,0,0],
    'marks':[80,70,100,50,0,0,0],
    'package':[10,7,14,2,0,0,0]
}

df=pd.DataFrame(student_dict)
df.set_index('name', inplace=True)

#with loc we can get 
```

df.loc[:, 'Age']

df.iloc[:, 1]

means → give me the column at position/index 1


```python

# whole row data present at index 0 
df.iloc[0]

## all the rows from index 0->3
print(df.iloc[0:4])

##we can also get data based on row and column

print(df.iloc[1,2])#=> row 1 col 2


## we can also user fancy indexing in dataframes
print(df.iloc[[0,4,5]])
##above will give rows 0 4 and 5
##same can be done using loc in df 
print(df.loc[['nitish','rupesh']])
```
