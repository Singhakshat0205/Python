# astype # between # clip # drop_duplicates # isnull # dropna # fillna # isin # apply # copy

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'name': ['Akshat', 'Rahul', 'Priya', 'Aman', 'Rahul'],
    'age': [23, 25, 22, np.nan, 25],
    'salary': [50000, 60000, 45000, 70000, 60000],
    'city': ['Pune', 'Delhi', 'Mumbai', 'Pune', 'Delhi']
})
```

```python
print(df)
```

```python
#astype() is used to change the data type of a Series.
df.loc[df['age'].isnull(), 'age']=25
print(df['age'].astype(int))

#for permanent changes we can do 
df['age']= df['age'].astype('int32')
print(df)
print(df['age'].dtype)
print(type(df['age']))
```

```python
#between() checks whether values lie between two boundaries.

df['age'].between(23,40)
```

```python
#clip() limits values to a specified range.

df['salary'].between(45000, 50000)
df.columns
```

```python
#The clip() method in pandas trims values outside a specified lower and upper boundary. Any values smaller than the lower threshold are increased to that threshold, and any values larger than the upper threshold are decreased to that threshold.

s= pd.Series([1,2,3,4,5,6,7,8,9])

s.clip(lower=3,upper=7, inplace=True)

print(s)

```

```python
#between() → check range
# clip() → restrict values to range
```

```python
##drop_duplicates()
#This removes duplicate rows
new_row = {
    'name': 'Akshat',
    'age': np.nan,
    'salary': 40000,
    'city': 'Pune'
}



df= pd.concat([df, pd.DataFrame([new_row])], ignore_index= True)


#df=df.drop_duplicates()


print(df)
```

```python
df
```

```python
##dropna() removes rows or columns containing missing values.
df.dropna(inplace= True)
print(df)
```

```python
#Instead of deleting missing values, we can fill them.
#using fillna

df.loc[df['name']=='Akshat','name']=np.nan
print(df)



df['name'].fillna('Akshat',)
#or
df= df['name'].fillna('Akshat')
print(df)
```

```python
##isin() checks whether values belong to a particular set/list.

df['city'].isin(['Pune'])
```

```python
# apply() allows you to apply a function to each value.
# df['salary']= df['salary'].apply(lambda x: x*2 if x > 23 else x)
# print(df)
import numpy as np
import math
s= pd.Series([2,3,4,5,5,6,7,8,8,8,8,8,8])

print(math.sqrt(9))
s= s.apply(lambda x:x**2 if x>2 else x**3)

s= s.apply(lambda x:int(math.sqrt(x)) if x%2==0 else x)

print(s)
```
