```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "player": ["Virat", "Rohit", "Virat", "Dhoni", "Rohit"],
    "team": ["India", "India", "India", "India", "India"],
    "city": ["Delhi", "Mumbai", "Delhi", "Chennai", "Delhi"],
    "runs": [80, 45, 100, np.nan, 60],
    "balls": [60, 30, 70, 10, 40],
    "result": ["Win", "Win", "Loss", "Win", "Loss"]
})

print(df)
```

```python
df.value_counts('player')
df.value_counts('city')

df.value_counts('result')
print(df['result'].value_counts())
```

```python
df['player'].unique()
```

```python
#we can also count the number of unique values

print(df['player'].nunique())
```

```python
# by default unique does not count NaN values

print(df['runs'].nunique())
print(df['runs'].nunique(dropna=False))

df['runs'].nunique()==df['runs'].nunique(dropna= False)
```

```python
## for sorting the dataframes

df.sort_values(by='runs',ascending= False)
```

```python
# we can get the ranks with the help of sort_values and rank functions

df['runs_rank']= df['runs'].rank(ascending= False)

print(df)
```

```python
##sort_index

#below will sort the dataframe based on column index
# the columns are going to be sort based on alphabetical order
df.sort_index(axis= 1)
##

print(df)
##below is row wise sort_index
df.sort_index(axis=0)
print(df)

```

```python
#wew can also mark any columns as our index column

df.set_index('player', inplace= True)
```

```python
# we can also set multiple columns as our index
df.set_index(["runs", "city"])
```

```python
# we can also reset the index that we had given

df.reset_index()
```

```python
# we can also check for null values

df.isnull()

#if we want the sum of all the null values
df.isnull().sum()

#above will give us the number of null values per column
```

```python
#notnull

df['runs'].notnull()
```

```python
## fill the na values
df['runs']= df['runs'].fillna(df['runs'].mean())

print(df)
```
