```python
def fact(n):
    if n==2:
        return 2

    return n*fact(n-1)

fact(30)
```

```python
def revArr(arr,i,j):
    if i>=j:
        return arr

    arr[i],arr[j]=arr[j],arr[i]
    return revArr(arr,i+1,j-1)

arr=[1,2,3,4]
print(revArr(arr,0,3))

print(arr)
    
```

```python
def palind(st, i, j):

    if i>j:
        return True
    return palind(st,i+1,j-1) and st[i]==st[j]



palind('abba',0,len('abba')-1)
```
