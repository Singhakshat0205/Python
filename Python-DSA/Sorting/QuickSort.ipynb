```python
def quick(arr, low, high):

    if low < high:
        partind = func(arr, low, high)

        quick(arr, low, partind - 1)
        quick(arr, partind + 1, high)


def func(arr, low, high):

    i, j = low, high
    pivot = arr[low]

    while i < j:

        while i <= high and arr[i] <= pivot:
            i += 1

        while j >= low and arr[j] > pivot:
            j -= 1

        if i < j:
            arr[i], arr[j] = arr[j], arr[i]

    arr[low], arr[j] = arr[j], arr[low]

    return j


arr = [3, 2, 1]

quick(arr, 0, 2)

print(arr)
```
