```python
def Merge(arr, low, mid, high):
     
     i= low
     j=mid+1
     temp=[]
     while i<=mid and j<=high:
        if arr[i]>arr[j]:
            temp.append(arr[j])
            j+=1
            continue
        if arr[i]<=arr[j]:
            temp.append(arr[i])
            i+=1
            continue
        if arr[i]==arr[j]:
            temp.append(arr[i])
            temp.append(arr[j])
            i+=1
            j+=1
            continue

     while i<=mid:
          temp.append(arr[i])
          i+=1

     while j<=high:
          temp.append(arr[j])
          j+=1
     l=0
     for k in range(low, high+1):
        arr[k]=temp[l]
        l+=1



def MergeSort(arr,low,high):
   if low>=high:
      return
   
   mid= (low+high)//2
   MergeSort(arr,low, mid)
   MergeSort(arr,mid+1,high)

   Merge(arr, low, mid, high)

arr=[3,2,1]
MergeSort(arr,0,2)
print(arr)

import random

arrays = [
    random.sample(range(1, 101), 10)
    for _ in range(1000)
]

for lst in arrays:
    MergeSort(lst, 0, len(lst)-1)
    print(lst)
```
