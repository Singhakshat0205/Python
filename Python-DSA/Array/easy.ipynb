```python
# check if array  is sorted

print(arr==sorted(arr))
arr=[2,3,4,1,2]

def is_sorted(arr):
    return any(arr[i]>arr[i+1] for i in range(len(arr)-1)) ^ True

print(is_sorted(arr))

def is_sorted2(arr):
    return all(arr[i]<=arr[i+1] for i in range(len(arr)-1))
brr=[1,2,3,4,5]
print(is_sorted2(brr))
```
